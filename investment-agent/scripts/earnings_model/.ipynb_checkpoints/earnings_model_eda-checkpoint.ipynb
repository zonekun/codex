# 決算反応モデル — EDA

**目的**: 決算発表時の翌日株価騰落率を予測するモデルの特徴量を探索する

**データソース**:
- J-Quants `/fins/summary` — 決算実績・会社予想・配当・株式数
- BQ `STOCK.CONSENSUS` — コンセンサス（経常利益）
- BQ `STOCK.STOCK_PRICE_JQUANTS` — 株価（ラベル生成・時価総額計算）
- BQ `STOCK.STOCK_CODE_LIST` — セクター・市場区分
- BQ `STOCK.v_fin_summary_actual_for_q_on_q` — 前年同期実績（単独四半期値）

```python
%matplotlib inline
import os, sys
from pathlib import Path
from datetime import datetime, timedelta, date
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# SSL workaround（BQ ローカル接続用）
import urllib3, requests as _req
from requests.adapters import HTTPAdapter as _HA
urllib3.disable_warnings()
class _NoVerify(_HA):
    def send(self, req, **kw): kw['verify'] = False; return super().send(req, **kw)
_orig = _req.Session.__init__
def _p(self, *a, **kw): _orig(self, *a, **kw); self.mount('https://', _NoVerify()); self.verify = False
_req.Session.__init__ = _p

# パス設定
PROJECT_ROOT = Path(r'C:\gdrive\claude\investment-agent')
sys.path.insert(0, str(PROJECT_ROOT / 'scripts'))

from dotenv import load_dotenv
load_dotenv(PROJECT_ROOT / '.env')

from google.cloud import bigquery
from google.oauth2 import service_account
from jquants_common import jquants_get

# J-Quants ヘッダー
JQUANTS_HEADERS = {'x-api-key': os.environ['JQUANTS_API_KEY']}

# BigQuery クライアント
KEY_FILE = str(PROJECT_ROOT / 'keys' / 'gcp-service-account.json')
creds = service_account.Credentials.from_service_account_file(KEY_FILE)
bq = bigquery.Client(credentials=creds, project='gmailpj-357912')

print('Setup OK')
```

```python
# ── 設定 ──────────────────────────────────────────────────────
# 対象期間（決算発表日）
DATE_FROM = '20260330'   # ← 開始日 (YYYYMMDD)
DATE_TO   = '20260330'   # ← 終了日 (YYYYMMDD)

# pandas 表示設定
pd.set_option('display.max_columns', 50)
pd.set_option('display.float_format', '{:,.2f}'.format)
```

## 1. fin-summary 取得

- **`USE_BQ=True`（デフォルト）**: BQ `fin_summary` から取得 — モデル構築・過去データ分析用
- **`USE_BQ=False`**: J-Quants API から取得 — BQ未収録の最新データ用

```python
USE_BQ = True   # True=BQ（過去データ・モデル構築）/ False=J-Quants API（BQ未収録の最新データ）


def fetch_fin_summary_from_bq(date_from: str, date_to: str) -> pd.DataFrame:
    """BQ fin_summary から指定期間の決算データを取得し、API短縮名カラムで返す.

    同一銘柄・同一四半期に修正決算（DocType違い）が存在する場合は
    最新の DISCLOSED_DATE を優先（修正決算を使用）。
    ※ パーティションキー: (LOCAL_CODE, TYPE_OF_CURRENT_PERIOD, CURRENT_FISCAL_YEAR_END_DATE)
    """
    d_from = date_from[:4] + '-' + date_from[4:6] + '-' + date_from[6:]
    d_to   = date_to[:4]   + '-' + date_to[4:6]   + '-' + date_to[6:]
    df = bq.query(f"""
        WITH ranked AS (
            SELECT
                DISCLOSED_DATE,
                DISCLOSED_TIME,
                LOCAL_CODE,
                DISCLOSURE_NUMBER,
                TYPE_OF_DOCUMENT,
                TYPE_OF_CURRENT_PERIOD,
                CURRENT_FISCAL_YEAR_END_DATE,
                NET_SALES,
                OPERATING_PROFIT,
                ORDINARY_PROFIT,
                PROFIT,
                EARNINGS_PER_SHARE,
                TOTAL_ASSETS,
                EQUITY,
                CASH_FLOWS_FROM_OPERATING_ACTIVITIES,
                RESULT_DIVIDEND_PER_SHARE_ANNUAL,
                FORECAST_DIVIDEND_PER_SHARE_ANNUAL,
                FORECAST_PAYOUT_RATIO_ANNUAL,
                FORECAST_NET_SALES,
                FORECAST_OPERATING_PROFIT,
                FORECAST_ORDINARY_PROFIT,
                FORECAST_PROFIT,
                NEXT_YEAR_FORECAST_NET_SALES,
                NEXT_YEAR_FORECAST_OPERATING_PROFIT,
                NEXT_YEAR_FORECAST_ORDINARY_PROFIT,
                NEXT_YEAR_FORECAST_PROFIT,
                NUMBER_OF_ISSUED_AND_OUTSTANDING_SHARES_AT_THE_END_OF_FISCAL_YEAR_INCLUDING_TREASURY_STOCK,
                NUMBER_OF_TREASURY_STOCK_AT_THE_END_OF_FISCAL_YEAR,
                AVERAGE_NUMBER_OF_SHARES,
                -- 同一銘柄・同一四半期で複数開示がある場合、最新を1位にする（修正決算優先）
                ROW_NUMBER() OVER (
                    PARTITION BY LOCAL_CODE,
                                 TYPE_OF_CURRENT_PERIOD,
                                 CURRENT_FISCAL_YEAR_END_DATE
                    ORDER BY DISCLOSED_DATE DESC
                ) AS rn
            FROM `gmailpj-357912.STOCK.fin_summary`
            WHERE DISCLOSED_DATE BETWEEN '{d_from}' AND '{d_to}'
        )
        SELECT
            DISCLOSED_DATE                AS DiscDate,
            DISCLOSED_TIME                AS DiscTime,
            LOCAL_CODE                    AS Code,
            DISCLOSURE_NUMBER             AS DiscNo,
            TYPE_OF_DOCUMENT              AS DocType,
            TYPE_OF_CURRENT_PERIOD        AS CurPerType,
            CURRENT_FISCAL_YEAR_END_DATE  AS CurFYEn,
            NET_SALES                     AS Sales,
            OPERATING_PROFIT              AS OP,
            ORDINARY_PROFIT               AS OdP,
            PROFIT                        AS NP,
            EARNINGS_PER_SHARE            AS EPS,
            TOTAL_ASSETS                  AS TA,
            EQUITY                        AS Eq,
            CASH_FLOWS_FROM_OPERATING_ACTIVITIES AS CFO,
            RESULT_DIVIDEND_PER_SHARE_ANNUAL     AS DivAnn,
            FORECAST_DIVIDEND_PER_SHARE_ANNUAL   AS FDivAnn,
            FORECAST_PAYOUT_RATIO_ANNUAL         AS FPayoutRatioAnn,
            FORECAST_NET_SALES            AS FSales,
            FORECAST_OPERATING_PROFIT     AS FOP,
            FORECAST_ORDINARY_PROFIT      AS FOdP,
            FORECAST_PROFIT               AS FNP,
            NEXT_YEAR_FORECAST_NET_SALES        AS NxFSales,
            NEXT_YEAR_FORECAST_OPERATING_PROFIT AS NxFOP,
            NEXT_YEAR_FORECAST_ORDINARY_PROFIT  AS NxFOdP,
            NEXT_YEAR_FORECAST_PROFIT           AS NxFNp,
            NUMBER_OF_ISSUED_AND_OUTSTANDING_SHARES_AT_THE_END_OF_FISCAL_YEAR_INCLUDING_TREASURY_STOCK AS ShOutFY,
            NUMBER_OF_TREASURY_STOCK_AT_THE_END_OF_FISCAL_YEAR AS TrShFY,
            AVERAGE_NUMBER_OF_SHARES      AS AvgSh
        FROM ranked
        WHERE rn = 1
        ORDER BY DiscDate, Code
    """).to_dataframe()
    df['DiscDate'] = pd.to_datetime(df['DiscDate']).dt.date
    print(f'BQ取得: {len(df)} 件 ({d_from} ～ {d_to})  ※修正決算優先で重複除去済み')
    return df


def fetch_fin_summary_from_api(date_from: str, date_to: str) -> pd.DataFrame:
    """J-Quants API から指定期間の決算データを取得して返す（BQ未収録の最新データ用）.

    最新決算のため修正決算は存在しない前提。重複除去不要。
    """
    d_from = datetime.strptime(date_from, '%Y%m%d')
    d_to   = datetime.strptime(date_to,   '%Y%m%d')
    all_records = []
    d = d_from
    while d <= d_to:
        recs = jquants_get('/fins/summary', {'date': d.strftime('%Y%m%d')}, JQUANTS_HEADERS)
        if recs:
            print(f'{d:%Y-%m-%d}: {len(recs)} 件')
            all_records.extend(recs)
        d += timedelta(days=1)
    if not all_records:
        return pd.DataFrame()
    df = pd.DataFrame(all_records).replace('', np.nan)
    df['Code']     = df['Code'].str[:4]
    df['DiscDate'] = pd.to_datetime(df['DiscDate']).dt.date
    str_cols = {'Code','DiscNo','DocType','CurPerType','DiscTime',
                'CurPerSt','CurPerEn','CurFYSt','CurFYEn','NxtFYSt','NxtFYEn',
                'MatChgSub','SigChgInC','ChgByASRev','ChgNoASRev','ChgAcEst','RetroRst'}
    for col in df.columns:
        if col not in str_cols and col != 'DiscDate':
            df[col] = pd.to_numeric(df[col], errors='ignore')
    return df


if USE_BQ:
    df_raw = fetch_fin_summary_from_bq(DATE_FROM, DATE_TO)
else:
    df_raw = fetch_fin_summary_from_api(DATE_FROM, DATE_TO)

print(f'合計: {len(df_raw)} 件, カラム数: {len(df_raw.columns)}')
df_raw.head(3)
```

```python
# DocType（開示書類種別）の分布を確認
print(df_raw['DocType'].value_counts())
print()
# CurPerType（四半期種別）の分布
print(df_raw['CurPerType'].value_counts())
```

```python
# ── フィルタリング ──────────────────────────────────────────────
# 連結決算のみ（非連結 NC系 は補完用に残す）
# DocType: 'FY' = 通期, '1Q'/'2Q'/'3Q' = 四半期, etc.
# TODO: DocType の値を確認して必要に応じて絞り込む
df_earn = df_raw.copy()
print(f'対象件数: {len(df_earn)}')
df_earn[['Code', 'DiscDate', 'DiscTime', 'DocType', 'CurPerType', 'Sales', 'OP', 'OdP', 'NP']].head(10)
```

## 2. BQ データ取得

```python
# ── 2-1. セクターマスタ ────────────────────────────────────────
df_sector = bq.query("""
    SELECT
        TICKER,
        STOCK_NAME,
        MARKET_CATEGORY,
        INDUSTRY_33_CODE,
        INDUSTRY_33_CATEGORY,
        SIZE_CODE,
        SIZE_CATEGORY
    FROM `gmailpj-357912.STOCK.STOCK_CODE_LIST`
""").to_dataframe()
print(f'銘柄マスタ: {len(df_sector)} 件')
df_sector.head(3)
```

```python
# ── 2-2. コンセンサス（発表日以前の最新1件 / 銘柄別）────────────
# CONSENSUS は DATAAT カラム（DATE型）
tickers = df_earn['Code'].unique().tolist()
disc_dates = [str(d) for d in df_earn['DiscDate'].unique()]
max_disc_date = max(disc_dates)

df_consensus = bq.query(f"""
    WITH ranked AS (
        SELECT
            TICKER,
            DATAAT,
            FY,
            QUARTER,
            PROFIT,
            ROW_NUMBER() OVER (PARTITION BY TICKER ORDER BY DATAAT DESC) AS rn
        FROM `gmailpj-357912.STOCK.CONSENSUS`
        WHERE DATAAT <= '{max_disc_date}'
    )
    SELECT TICKER, DATAAT, FY, QUARTER, PROFIT AS CONSENSUS_PROFIT
    FROM ranked
    WHERE rn = 1
""").to_dataframe()
print(f'コンセンサス: {len(df_consensus)} 件')
df_consensus.head(3)
```

```python
# ── 2-3. 株価（発表日 + 翌営業日）────────────────────────────────
# ラベル: 翌日終値騰落率
# 時価総額: 発表日前日終値 × (ShOutFY - TrShFY)
date_from_sql = DATE_FROM[:4] + '-' + DATE_FROM[4:6] + '-' + DATE_FROM[6:]
# 前後3営業日分を取得して後でjoin
date_from_ext = (datetime.strptime(DATE_FROM, '%Y%m%d') - timedelta(days=5)).strftime('%Y-%m-%d')
date_to_ext   = (datetime.strptime(DATE_TO,   '%Y%m%d') + timedelta(days=5)).strftime('%Y-%m-%d')

df_price = bq.query(f"""
    SELECT
        DATE,
        TICKER,
        ADJ_OPEN,
        ADJ_CLOSE,
        VOLUME
    FROM `gmailpj-357912.STOCK.STOCK_PRICE_JQUANTS`
    WHERE DATE BETWEEN '{date_from_ext}' AND '{date_to_ext}'
      AND IS_PREFERRED = FALSE
    ORDER BY TICKER, DATE
""").to_dataframe()
df_price['DATE'] = pd.to_datetime(df_price['DATE']).dt.date
print(f'株価: {len(df_price)} 件')
df_price.head(3)
```

```python
# ── 2-4. 前年同期実績（Q単独値）
# v_fin_summary_actual_for_q_on_q ビュー: 累積P&L -> Q単独値に変換済み
_d_from = DATE_FROM[:4] + "-" + DATE_FROM[4:6] + "-" + DATE_FROM[6:]
_d_to   = DATE_TO[:4]   + "-" + DATE_TO[4:6]   + "-" + DATE_TO[6:]

df_yoy = bq.query(
    f"""
    SELECT
        LOCAL_CODE       AS Code,
        DISCLOSED_DATE   AS DiscDate,
        QUARTER,
        NET_SALES        AS Q_Sales,
        OPERATING_PROFIT AS Q_OP,
        ORDINARY_PROFIT  AS Q_OdP,
        PROFIT           AS Q_NP
    FROM `gmailpj-357912.STOCK.v_fin_summary_actual_for_q_on_q`
    WHERE DISCLOSED_DATE BETWEEN "{_d_from}" AND "{_d_to}"
      AND TYPE_OF_DOCUMENT LIKE "%Consolidated%"
    """
).to_dataframe()
df_yoy["DiscDate"] = pd.to_datetime(df_yoy["DiscDate"]).dt.date
print(f"Q単独実績: {len(df_yoy)} 件")
print(f"QUARTER分布:"); print(df_yoy["QUARTER"].value_counts())
df_yoy.head(3)
```

## 3. データ結合

```python
# ── 翌日株価を計算 ──────────────────────────────────────────────
# 各銘柄の取引日カレンダーを作成 → DiscDate の次の取引日を特定
df_price_sorted = df_price.sort_values(['TICKER', 'DATE'])
df_price_sorted['NEXT_DATE'] = df_price_sorted.groupby('TICKER')['DATE'].shift(-1)
df_price_sorted['NEXT_ADJ_CLOSE'] = df_price_sorted.groupby('TICKER')['ADJ_CLOSE'].shift(-1)
df_price_sorted['NEXT_ADJ_OPEN']  = df_price_sorted.groupby('TICKER')['ADJ_OPEN'].shift(-1)

# 翌日終値騰落率 / 翌日始値騰落率
df_price_sorted['LABEL_CLOSE_RETURN'] = (
    df_price_sorted['NEXT_ADJ_CLOSE'] / df_price_sorted['ADJ_CLOSE'] - 1
)
df_price_sorted['LABEL_OPEN_RETURN'] = (
    df_price_sorted['NEXT_ADJ_OPEN'] / df_price_sorted['ADJ_CLOSE'] - 1
)

df_price_label = df_price_sorted[[
    'TICKER', 'DATE', 'ADJ_CLOSE', 'VOLUME',
    'LABEL_CLOSE_RETURN', 'LABEL_OPEN_RETURN'
]].copy()
df_price_label.head(3)
```

```python
# ── 全テーブル結合
df = df_earn.copy()

df = df.merge(df_sector, left_on="Code", right_on="TICKER", how="left")

df = df.merge(
    df_consensus[["TICKER", "CONSENSUS_PROFIT", "DATAAT"]],
    left_on="Code", right_on="TICKER", how="left"
)

df = df.merge(
    df_price_label.rename(columns={"TICKER": "Code_price", "DATE": "DiscDate"}),
    left_on=["Code", "DiscDate"],
    right_on=["Code_price", "DiscDate"],
    how="left"
)

# Q単独実績（v_fin_summary_actual_for_q_on_q）
df = df.merge(
    df_yoy[["Code", "DiscDate", "Q_Sales", "Q_OP", "Q_OdP", "Q_NP"]],
    on=["Code", "DiscDate"],
    how="left"
)

print(f"結合後: {len(df)} 件, カラム数: {len(df.columns)}")
print("ラベル欠損率:"); print(df[["LABEL_CLOSE_RETURN", "LABEL_OPEN_RETURN"]].isnull().mean())
print("Q単独実績欠損率:"); print(df[["Q_Sales", "Q_OP", "Q_OdP", "Q_NP"]].isnull().mean())
```

## 4. 特徴量エンジニアリング

```python
def safe_ratio(numerator, denominator, clip: float = 5.0) -> pd.Series:
    denom = denominator.abs().replace(0, np.nan)
    return ((numerator - denominator) / denom).clip(-clip, clip)


REMAINING_Q = {"1Q": 3, "2Q": 2, "3Q": 1, "FY": 0}
remaining_q = df["CurPerType"].map(REMAINING_Q)
is_fy = df["CurPerType"] == "FY"

feat = pd.DataFrame(index=df.index)
feat["Code"]       = df["Code"]
feat["DiscDate"]   = df["DiscDate"]
feat["CurPerType"] = df["CurPerType"]
feat["DocType"]    = df["DocType"]

# 4-1. Q単独実績 vs 会社予想（Q単独ペース換算）
# 会社予想Qペース = (通期予想 - 当期累積実績) / 残りQ数
# 注: 前回発表時のFSalesが本来正確。現状は当期発表値で近似（TODOを参照）
implied_q_sales = (df["FSales"] - df["Sales"]) / remaining_q.replace(0, np.nan)
implied_q_op    = (df["FOP"]    - df["OP"])    / remaining_q.replace(0, np.nan)
implied_q_np    = (df["FNP"]    - df["NP"])    / remaining_q.replace(0, np.nan)

feat["surprise_sales_vs_forecast"] = np.where(
    is_fy,
    safe_ratio(df["Sales"],   df["FSales"]),
    safe_ratio(df["Q_Sales"], implied_q_sales)
)
feat["surprise_op_vs_forecast"] = np.where(
    is_fy,
    safe_ratio(df["OP"],   df["FOP"]),
    safe_ratio(df["Q_OP"], implied_q_op)
)
feat["surprise_np_vs_forecast"] = np.where(
    is_fy,
    safe_ratio(df["NP"],   df["FNP"]),
    safe_ratio(df["Q_NP"], implied_q_np)
)

# 4-2. 実績（累積）vs コンセンサス（累積） -> 累積同士でOK
# 注: 全QのCONSENSUSが揃っていない。Q単独変換はデータパターン確認後にTODO
feat["surprise_odp_vs_consensus"] = safe_ratio(df["OdP"], df["CONSENSUS_PROFIT"])

# 4-3. 翌期予想 vs 今期実績比（FY のみ有意）
feat["guidance_fy_sales"] = safe_ratio(df["NxFSales"], df["Sales"])
feat["guidance_fy_op"]    = safe_ratio(df["NxFOP"],    df["OP"])
feat["guidance_fy_np"]    = safe_ratio(df["NxFNp"],    df["NP"])

# 4-3b. 残りQ期間の会社予想ペース（Q1-Q3で有効）
feat["guidance_remaining_q_op"] = implied_q_op

# 4-4. 配当変化
feat["dividend_surprise"] = safe_ratio(df["DivAnn"], df["FDivAnn"], clip=3.0)

# 4-5. 時価総額（発表日終値 × 浮動株数）
float_shares = (df["ShOutFY"] - df["TrShFY"]).clip(lower=1)
feat["market_cap"] = (df["ADJ_CLOSE"] * float_shares).apply(
    lambda x: np.log1p(x) if pd.notna(x) and x > 0 else np.nan
)

feat["quarter"]     = df["CurPerType"]
feat["industry_33"] = df["INDUSTRY_33_CATEGORY"]
feat["size_cat"]    = df["SIZE_CATEGORY"]

feat["label_close_return"] = df["LABEL_CLOSE_RETURN"]
feat["label_open_return"]  = df["LABEL_OPEN_RETURN"]

print(f"特徴量DataFrame: {feat.shape}")
print("Q単独サプライズ欠損率:")
print(feat[["surprise_sales_vs_forecast","surprise_op_vs_forecast","surprise_np_vs_forecast"]].isnull().mean())
feat.head(5)
```

## 4-b. 特徴量 GCS 保存 / 読み込み

```python
import subprocess
from pathlib import Path

GCS_BUCKET    = 'gs://stock_data_1930932'
GCS_FEATURES  = f'{GCS_BUCKET}/earnings_model/features'
GCS_MODELS    = f'{GCS_BUCKET}/earnings_model/models'
GCS_REPORTS   = f'{GCS_BUCKET}/earnings_model/reports'
TMP_DIR       = Path(r'C:\tmp\earnings_model')


def save_features_to_gcs(df: pd.DataFrame, date_from: str, date_to: str) -> str:
    """特徴量DataFrameをCSVに保存してGCSにアップロードする.

    Args:
        df:        保存する特徴量DataFrame
        date_from: 期間開始 YYYYMMDD
        date_to:   期間終了 YYYYMMDD

    Returns:
        アップロード先 GCS パス
    """
    TMP_DIR.mkdir(parents=True, exist_ok=True)
    filename  = f'{date_from}_{date_to}.csv'
    tmp_path  = TMP_DIR / filename
    gcs_path  = f'{GCS_FEATURES}/{filename}'

    df.to_csv(tmp_path, index=False, encoding='utf-8')
    print(f'一時保存: {tmp_path}  ({tmp_path.stat().st_size / 1024:.1f} KB)')

    result = subprocess.run(
        ['gcloud', 'storage', 'cp', str(tmp_path), gcs_path],
        capture_output=True, text=True
    )
    if result.returncode != 0:
        print(f'[ERROR] GCSアップロード失敗:\n{result.stderr}')
        return ''

    tmp_path.unlink()   # 一時ファイル削除
    print(f'GCS保存完了: {gcs_path}')
    return gcs_path


def load_features_from_gcs(date_from: str, date_to: str) -> pd.DataFrame:
    """GCSから特徴量CSVをダウンロードして DataFrame で返す.

    Args:
        date_from: 期間開始 YYYYMMDD
        date_to:   期間終了 YYYYMMDD
    """
    TMP_DIR.mkdir(parents=True, exist_ok=True)
    filename = f'{date_from}_{date_to}.csv'
    tmp_path = TMP_DIR / filename
    gcs_path = f'{GCS_FEATURES}/{filename}'

    result = subprocess.run(
        ['gcloud', 'storage', 'cp', gcs_path, str(tmp_path)],
        capture_output=True, text=True
    )
    if result.returncode != 0:
        print(f'[ERROR] GCSダウンロード失敗:\n{result.stderr}')
        return pd.DataFrame()

    df = pd.read_csv(tmp_path, encoding='utf-8')
    df['DiscDate'] = pd.to_datetime(df['DiscDate']).dt.date
    tmp_path.unlink()
    print(f'GCS読み込み完了: {gcs_path}  ({len(df)} 件)')
    return df


def list_features_on_gcs() -> None:
    """GCS上の特徴量ファイル一覧を表示する."""
    result = subprocess.run(
        ['gcloud', 'storage', 'ls', f'{GCS_FEATURES}/'],
        capture_output=True, text=True
    )
    print(result.stdout if result.stdout else '(ファイルなし)')


# ── 保存実行 ──────────────────────────────────────────────────
# feat が作成済みの場合に実行（EDA後に保存でも可）
# save_features_to_gcs(feat, DATE_FROM, DATE_TO)

# ── 一覧確認 ──────────────────────────────────────────────────
list_features_on_gcs()
```

```python
# ── 5-2. 四半期別リターン分布 ────────────────────────────────────
feat.boxplot(column='label_close_return', by='quarter', figsize=(10, 5))
plt.suptitle('')
plt.title('四半期別 翌日終値騰落率')
plt.xlabel('四半期種別')
plt.ylabel('Return')
plt.axhline(0, color='red', linestyle='--')
plt.show()
```

```python
# ── 5-3. 因子 × ラベル 散布図 ──────────────────────────────────
factor_cols = [
    'surprise_np_vs_forecast',
    'surprise_odp_vs_consensus',
    'guidance_fy_np',
    'dividend_surprise',
    'market_cap',
]

fig, axes = plt.subplots(1, len(factor_cols), figsize=(18, 4))
for ax, col in zip(axes, factor_cols):
    sub = feat[[col, 'label_close_return']].dropna()
    ax.scatter(sub[col], sub['label_close_return'], alpha=0.3, s=10)
    ax.axhline(0, color='red', linestyle='--', linewidth=0.8)
    ax.axvline(0, color='gray', linestyle='--', linewidth=0.8)
    ax.set_xlabel(col)
    ax.set_ylabel('label_close_return')
    # 相関係数
    if len(sub) > 1:
        corr = sub.corr().iloc[0, 1]
        ax.set_title(f'r={corr:.3f} (n={len(sub)})')

plt.tight_layout()
plt.show()
```

```python
# ── 5-4. 単変量IC（情報係数）一覧 ────────────────────────────────
from scipy import stats

ic_results = []
for col in factor_cols:
    sub = feat[[col, 'label_close_return']].dropna()
    if len(sub) < 10:
        continue
    corr, pval = stats.pearsonr(sub[col], sub['label_close_return'])
    spearman_r, spearman_p = stats.spearmanr(sub[col], sub['label_close_return'])
    ic_results.append({
        'factor': col,
        'n': len(sub),
        'pearson_r': corr,
        'pearson_p': pval,
        'spearman_r': spearman_r,
        'spearman_p': spearman_p,
    })

df_ic = pd.DataFrame(ic_results).sort_values('spearman_r', ascending=False)
df_ic
```

```python
# ── 5-5. セクター別リターン中央値 ────────────────────────────────
sector_ret = (
    feat.groupby('industry_33')['label_close_return']
    .agg(['median', 'mean', 'count'])
    .sort_values('median', ascending=False)
)
sector_ret.plot(y='median', kind='bar', figsize=(14, 4),
                title='セクター別 翌日終値騰落率（中央値）')
plt.axhline(0, color='red', linestyle='--')
plt.tight_layout()
plt.show()
sector_ret
```
